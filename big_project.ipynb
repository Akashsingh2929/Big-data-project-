```python
users = spark.table("fraud_catalog.fraud_schema.users")
cards = spark.table("fraud_catalog.fraud_schema.cards")
transactions = spark.table("fraud_catalog.fraud_schema.transaction")
display(users)
display(cards)
display(transactions)
```

```python
print("Users:", users.columns)
print("Cards:", cards.columns)
print("Transactions:", transactions.columns)
```

```python
user_cards = users.join(cards, on="user_id", how="inner")
```

```python
df = user_cards.join(transactions, on=["user_id", "card_id"], how="inner")
```

```python

display(df)
```

```python
from pyspark.sql.functions import col, when, hour, to_timestamp, dayofweek, month
```

```python
from pyspark.sql.functions import col

# Remove duplicates
df = df.dropDuplicates()

# Remove null values
df = df.dropna()

# Fix data types
df = df.withColumn("amount", col("amount").cast("double"))

df = df.withColumn("card_limit", col("card_limit").cast("double"))
df = df.withColumn("age", col("age").cast("string"))
```

```python
from pyspark.sql.functions import to_timestamp

# Convert the 'timestamp' column from String to actual Timestamp type
df = df.withColumn("timestamp", to_timestamp("timestamp", "M/d/yyyy H:mm"))
```

```python
# Fraud vs Non-Fraud count
df.groupBy("fraud_flag").count().show()

from pyspark.sql import functions as F

df.groupBy("location") \
  .agg(F.avg("fraud_flag").alias("avg_fraud")) \
  .orderBy(F.col("avg_fraud").desc()) \
  .show()

df.groupBy("device_type").avg("fraud_flag").show()
```

```python

# Convert timestamp
df = df.withColumn("timestamp", to_timestamp(col("timestamp")))

# Time features
df = df.withColumn("transaction_hour", hour("timestamp"))
df = df.withColumn("transaction_day", dayofweek("timestamp"))
df = df.withColumn("transaction_month", month("timestamp"))
```

```python
df = df.withColumn("timestamp", to_timestamp("timestamp", "M/d/yyyy H:mm"))
# Then save the DataFrame
df.write.mode("overwrite").saveAsTable("transactions_master")
```

```python
from pyspark.sql.functions import month

transactions = transactions.withColumn("transaction_month", month(to_timestamp("timestamp", 'M/d/yyyy H:mm')))

months = transactions.select("timestamp").distinct().collect()

for m in months:
    df = transactions.filter(col("transaction_month") == m[0])
    
    batch_join = df.join(users, "user_id").join(cards, "card_id")
    
    batch_join.write.mode("append").saveAsTable("monthly_fraud_results")
```

```python

from pyspark.sql.functions import col, when

df = df.withColumn(
    "fraud_detected",
    when(
        (col("amount") > 50000) |
        (col("is_foreign_transaction") == 1) |
        (col("transaction_hour") < 4) |
        (col("card_status") == "Blocked"),
        1
    ).otherwise(0)
)
```

```python
fraud_by_location = df.filter(col("fraud_flag") == 1).groupBy("location").count()
fraud_by_location.write.mode("overwrite").saveAsTable("fraud_by_location")

high_risk_users = df.filter(col("fraud_flag") == 1).groupBy("user_id").count()
high_risk_users.write.mode("overwrite").saveAsTable("high_risk_users")
```
